```python
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['Fare'] < lower) | (df['Fare'] > upper)]
print(outliers)
```

```python
df = df[(df['Fare'] >= lower) & (df['Fare'] <= upper)]
```

```python
df['Fare'] = df['Fare'].clip(lower, upper)
```
